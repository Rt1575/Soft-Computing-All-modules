# Experiment 1: Basic Working of an Artificial Neuron

**Aim:** To understand and implement the basic working of an artificial neuron using Python.

An artificial neuron computes a weighted sum of its inputs, adds a bias, and passes the result through an activation function to produce an output.

```python
import numpy as np

def artificial_neuron(inputs, weights, bias, activation="sigmoid"):
    """Simulate a single artificial neuron."""
    z = np.dot(inputs, weights) + bias   # weighted sum + bias

    if activation == "sigmoid":
        output = 1 / (1 + np.exp(-z))
    elif activation == "relu":
        output = max(0, z)
    else:
        output = z  # linear

    return z, output
```

```python
# Example: neuron with 3 inputs
inputs = np.array([0.5, 0.3, 0.2])
weights = np.array([0.4, 0.7, 0.1])
bias = 0.1

z, output = artificial_neuron(inputs, weights, bias, activation="sigmoid")
print(f"Weighted sum (z): {z:.4f}")
print(f"Neuron output (sigmoid): {output:.4f}")
```

## Conclusion

A single artificial neuron was implemented from scratch, showing how inputs are combined with weights and a bias, then squashed by an activation function to produce an output.
