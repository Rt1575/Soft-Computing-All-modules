# Experiment 5: Multilayer Perceptron (MLP) using Keras/TensorFlow

**Aim:** To create a Multilayer Perceptron (MLP) with input, hidden, and output layers using Keras/TensorFlow.

An MLP consists of an input layer, one or more hidden layers, and an output layer, each fully connected to the next.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential([
    Input(shape=(4,)),                 # input layer: 4 features
    Dense(8, activation="relu"),       # hidden layer 1
    Dense(4, activation="relu"),       # hidden layer 2
    Dense(1, activation="sigmoid"),    # output layer (binary classification)
])

model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

## Conclusion

A Multilayer Perceptron with two hidden layers was built using Keras, illustrating how the input, hidden, and output layers are stacked into a Sequential model.
