# Experiment 7: Basic Concept of Backpropagation

**Aim:** To understand and implement the basic concept of Backpropagation for training a neural network.

Backpropagation computes gradients of the loss with respect to every weight by applying the chain rule backward through the network, here for a small 2-2-1 network trained on XOR.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])   # XOR

np.random.seed(0)
W1 = np.random.uniform(-1, 1, (2, 2))   # input -> hidden
b1 = np.zeros((1, 2))
W2 = np.random.uniform(-1, 1, (2, 1))   # hidden -> output
b2 = np.zeros((1, 1))
lr = 0.5
```

```python
for epoch in range(5000):
    # ---- Forward pass ----
    hidden_in = np.dot(X, W1) + b1
    hidden_out = sigmoid(hidden_in)
    final_in = np.dot(hidden_out, W2) + b2
    final_out = sigmoid(final_in)

    # ---- Backward pass (backpropagation) ----
    error = y - final_out
    d_output = error * sigmoid_derivative(final_out)

    error_hidden = np.dot(d_output, W2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_out)

    # ---- Update weights and biases ----
    W2 += lr * np.dot(hidden_out.T, d_output)
    b2 += lr * np.sum(d_output, axis=0, keepdims=True)
    W1 += lr * np.dot(X.T, d_hidden)
    b1 += lr * np.sum(d_hidden, axis=0, keepdims=True)

print("Predictions after training (rounded):")
for x, pred in zip(X, final_out):
    print(f"{x} -> {pred[0]:.3f} -> {round(pred[0])}")
```

## Conclusion

Backpropagation was implemented manually for a 2-2-1 network and successfully learned the non-linearly separable XOR function.
