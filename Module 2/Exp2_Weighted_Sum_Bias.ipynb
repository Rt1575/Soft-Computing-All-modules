# Experiment 2: Weighted Sum of Inputs and Bias

**Aim:** To implement the weighted sum of inputs and bias used in an artificial neuron using Python.

The weighted sum (also called the net input) is computed as z = w1*x1 + w2*x2 + ... + wn*xn + b, before any activation function is applied.

```python
import numpy as np

def weighted_sum(inputs, weights, bias):
    """Compute z = w.x + b for a set of inputs."""
    return np.dot(inputs, weights) + bias
```

```python
# Demonstrate with multiple example input vectors
X = np.array([
    [1.0, 0.5, -0.3],
    [0.2, 0.8,  0.1],
    [-0.5, 0.4, 0.9],
])
weights = np.array([0.6, -0.2, 0.9])
bias = 0.05

for i, x in enumerate(X, start=1):
    z = weighted_sum(x, weights, bias)
    print(f"Input {i}: {x} -> weighted sum z = {z:.4f}")
```

## Conclusion

The weighted sum of inputs plus bias was computed for several input vectors, forming the linear pre-activation stage that every neuron relies on.
