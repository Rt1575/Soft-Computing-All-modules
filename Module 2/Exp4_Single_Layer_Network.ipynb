# Experiment 4: Simple Single-Layer Neural Network

**Aim:** To create a simple single-layer neural network using Python.

A single-layer network (perceptron) learns a linear decision boundary. Here it is trained on the AND logic gate.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Training data: AND gate
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[0],[0],[1]])

np.random.seed(1)
weights = np.random.uniform(-1, 1, (2, 1))
bias = np.random.uniform(-1, 1, (1,))
lr = 0.5
```

```python
# Train for a fixed number of epochs
for epoch in range(2000):
    z = np.dot(X, weights) + bias
    output = sigmoid(z)

    error = y - output
    d_output = error * output * (1 - output)

    weights += lr * np.dot(X.T, d_output)
    bias += lr * np.sum(d_output)

print("Final predictions (rounded):")
for x, pred in zip(X, output):
    print(f"{x} -> {pred[0]:.3f} -> {round(pred[0])}")
```

## Conclusion

A single-layer network was trained on the AND gate and correctly learned the linearly separable mapping between inputs and output.
