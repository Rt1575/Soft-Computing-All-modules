# Experiment 6: Gradient Descent for Weight Updates

**Aim:** To apply Gradient Descent for updating the weights of a simple neural network during training.

Gradient Descent updates each weight in the direction that reduces the loss: w = w - learning_rate * (dLoss/dw).

```python
import numpy as np

# Simple linear neuron: y_hat = w*x + b, trained with Mean Squared Error
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)   # true relation: y = 2x + 1

w, b = 0.0, 0.0
lr = 0.01
losses = []

for epoch in range(200):
    y_pred = w * X + b
    error = y_pred - y
    loss = np.mean(error ** 2)
    losses.append(loss)

    dw = np.mean(2 * error * X)   # gradient wrt w
    db = np.mean(2 * error)       # gradient wrt b

    w -= lr * dw
    b -= lr * db

print(f"Learned weight: {w:.4f}, bias: {b:.4f}")
print(f"Final loss: {losses[-1]:.6f}")
```

## Conclusion

Gradient Descent was used to iteratively update a neuron's weight and bias, converging close to the true underlying relationship y = 2x + 1.
